## Section 5: Saliency Map Computation (LRP & Integrated Gradients)

Two complementary XAI methods are used to obtain pixel-level attribution maps:

- **LRP (Layer-wise Relevance Propagation):** Backpropagates a one-hot gradient to estimate relevance.
- **Integrated Gradients (IG):** Integrates gradients along the path from a zero baseline to the input.

```python
# ── A Function to Calculate LRP ─────────────────────────────────────────────────

def compute_lrp(model, input_tensor, target_class):
    """
    Compute an LRP-style relevance map using gradient backpropagation.

    Args:
        model        : Trained PyTorch model
        input_tensor : Input image tensor with requires_grad=True, shape [1,C,H,W]
        target_class : Integer class index to explain

    Returns:
        relevance (Tensor): 2-D relevance map of shape [H, W]
    """
    model.zero_grad()  # Clear gradients from the previous pass

    # Create a one hot gradient vector
    one_hot = torch.zeros_like(output)
    one_hot[0, target_class] = 1

    # Backpropagate the one hot signal to compute input space gradents
    output.backward(gradient=one_hot)

    # Retrieve gradients w.r.t. the input tensor, then collapse colour channels
    # by summing absolute values → shape [H, W]
    relevance = input_tensor.grad.clone().detach()
    relevance = relevance.abs().sum(dim=1)[0]  # Sum over dimensions

    return relevance
```

```python
# Compute the LRP heatmap for the predicted class
lrp_map = compute_lrp(model, input_tensor, prediction).cpu().numpy()

# Min-max normalise to [0, 1] for consistent visualisation
# A small epsilon is added (1e-8) to prevent division by zero when the map is constant
lrp_map = (lrp_map - lrp_map.min()) / (lrp_map.max() - lrp_map.min() + 1e-8)
```

```python
# ── Integrated Gradients ─────────────────────────────────────────────────
ig = IntegratedGradients(model)

# Baseline: initilizing a new tensor with all zeros
baseline = torch.zeros_like(input_tensor)

# Compute IG attributions for the predicted class
# return_convergence_delta=True also returns the approximation error
attributions, _ = ig.attribute(input_tensor, baseline, target=prediction,
                               return_convergence_delta=True)

# Post-process: squeeze batch dim, move to numpy, convert [C,H,W] → [H,W,C]
ig_map = attributions.squeeze().detach().numpy()
ig_map = np.transpose(ig_map, (1, 2, 0))

# Collapse to single channel by averaging over colour channels
ig_map = np.mean(ig_map, axis=2)

# Keep only positive attributions (negative = suppresses the prediction)
ig_map = np.maximum(ig_map, 0)

# Normalise to [0, 1]
ig_map /= ig_map.max() + 1e-8
```

## Section 6: Thresholding & Computing the LRP ∩ IG Intersection

Convert continuous heatmaps to binary masks at a high-percentile threshold, then intersect them to retain only pixels that *both* methods agree are important.

```python
def threshold_map(saliency_map, percentile=80):
    """
    Binarise a saliency map by keeping pixels above a given percentile.

    Args:
        saliency_map : 2-D numpy array of relevance scores
        percentile   : Threshold percentile (default is 80, taking in account: top 20% of pixels)

    Returns:
        Binary masked image with 1 for salient pixels, 0 otherwise
    """
    threshold = np.percentile(saliency_map, percentile)
    return (saliency_map >= threshold).astype(np.uint8)
```

```python
# Threshold both maps at the 95th percentile, top 5% most relevant pixels (default is 80)
lrp_thresh = threshold_map(lrp_map, percentile=95)
ig_thresh  = threshold_map(ig_map,  percentile=95)

# Intersection: pixels that are salient in BOTH methods
# Using bitwise AND ensures only pixels agreed upon by LRP AND IG are kept
intersection = (lrp_thresh & ig_thresh).astype(np.uint8)

# ── Visualisation ────────────────────────────────────────────────────────
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(lrp_map, cmap='gray')
plt.title("LRP Heatmap")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(ig_map, cmap='gray')
plt.title("IG Heatmap")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(intersection, cmap='gray')
plt.title("Common Pixels (IG ∩ LRP)")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow((lrp_thresh + ig_thresh), cmap='gray')
plt.title("Overlay of LRP + IG Thresholds")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Section 7: Clustering Salient Pixels into Causal Concepts

Group the intersected salient pixels into spatially coherent clusters using **Agglomerative (Hierarchical) Clustering**. Each cluster represents a *causal concept* — a semantically meaningful image region.

```python
def cluster_agglomerative(binary_map, n_clusters=10, linkage='ward'):
    """
    Clusters active pixels in a binary map into n_clusters groups.

    Args:
        binary_map : 2-D uint8 array where 1 = salient pixel
        n_clusters : Number of clusters to form
        linkage    : Linkage criterion for AgglomerativeClustering (here ward criterion is used)


    Returns:
        labeled_map (int array): Same shape as binary_map;
                                 cluster id for salient pixels, -1 for background
    """
    # Extract (row, col) coordinates of all salient pixels
    coords = np.argwhere(binary_map == 1)

    # For Edge case: no salient pixels found → return all zero map
    if coords.shape[0] == 0:
        return np.zeros_like(binary_map, dtype=int)

    # Fit agglomerative clustering on pixel coordinates
    agg = AgglomerativeClustering(n_clusters=n_clusters, linkage=linkage)
    agg.fit(coords)

    # Build the output label map: fill background with -1, then assign cluster ids
    labels = agg.labels_
    labeled_map = np.full(binary_map.shape, -1, dtype=int)
    for i, coord in enumerate(coords):
        labeled_map[coord[0], coord[1]] = labels[i]

    return labeled_map
```

```python
# Cluster the intersection mask into 10 causal concept regions
agg_labels = cluster_agglomerative(intersection, n_clusters=10, linkage='ward')

# Mask background pixels (-1) so they render as white (transparent) in the colormap
masked_labels = np.ma.masked_where(agg_labels == -1, agg_labels)

# Use a qualitative colormap with 10 distinct colours — one per cluster
cmap = plt.get_cmap('tab10', 10)
cmap.set_bad(color='white')  # Background pixels → white

# Visualise the clustered causal concepts
fig, ax = plt.subplots(figsize=(12, 6))
im = ax.imshow(masked_labels, cmap=cmap)
ax.set_title("Agglomerative Clustering (10 clusters)")
ax.axis("off")
plt.tight_layout()
plt.show()
```

## Section 8: Image Utility Functions

Helper functions to reverse ImageNet normalisation and draw cluster bounding boxes on the original image.

```python
def unnormalize(img_tensor, mean, std):
    """
    Reverse ImageNet normalisation to recover displayable pixel values.

    Args:
        img_tensor : Tensor of shape [C, H, W]
        mean       : Per-channel mean used during normalisation
        std        : Per-channel std  used during normalisation

    Returns:
        img_tensor : Denormalised tensor of shape [C, H, W]
    """
    img_tensor = img_tensor.clone()  # Work on a copy to avoid modifying the original
    for i in range(img_tensor.shape[0]):
        img_tensor[i] = img_tensor[i] * std[i] + mean[i]  # x = x_norm * std + mean
    return img_tensor

def unnormalize_and_to_np(img_tensor, mean, std):
    """
    Convenience wrapper: unnormalise a tensor and return a uint8 numpy array.

    Returns:
        numpy array of shape [H, W, C], dtype uint8, values in [0, 255]
    """
    x = img_tensor.clone()
    for i in range(x.shape[0]):
        x[i] = x[i] * std[i] + mean[i]
    x_np = x.detach().cpu().numpy().transpose(1, 2, 0)  # [C,H,W] → [H,W,C]
    x_np = np.clip(x_np, 0, 1)                          # Clamp stray values to [0,1]
    return (x_np * 255).astype(np.uint8)
```

```python
# Recover the displayable image from the normalised input tensor
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]

# unnormalize returns a tensor; convert to numpy for matplotlib
transformed_img    = unnormalize(input_tensor[0], mean, std)
transformed_img_np = transformed_img.detach().cpu().numpy().transpose(1, 2, 0)
transformed_img_np = np.clip(transformed_img_np, 0, 1)  # Ensure [0,1] range for imshow
```

```python
def draw_cluster_boundaries(image_np, labeled_map, title="Cluster Boundaries", skip_noise=True):
    """
    Overlay axis-aligned bounding boxes for each cluster on an image.

    Args:
        image_np    : [H, W, C] float array in [0, 1], the background image
        labeled_map : [H, W] int array with cluster ids; -1 = background/noise
        title       : Plot title string
        skip_noise  : If True, skip cluster id -1 (background pixels)
    """
    fig, ax = plt.subplots(1, figsize=(8, 8))
    ax.imshow(image_np)

    clusters = np.unique(labeled_map)
    for cl in clusters:
        if skip_noise and cl == -1:  # Skip background label
            continue

        # Get bounding box from min/max row,col of all pixels in this cluster
        coords = np.argwhere(labeled_map == cl)
        if coords.size == 0:
            continue
        min_row, min_col = coords.min(axis=0)
        max_row, max_col = coords.max(axis=0)
        width  = max_col - min_col
        height = max_row - min_row

        # Draws a red rectangle around the cluster
        rect = patches.Rectangle(
            (min_col, min_row), width, height,
            linewidth=2, edgecolor='red', facecolor='none'
        )
        ax.add_patch(rect)
        # Label the box with the cluster ID
        ax.text(min_col, min_row, f"{cl}", color='yellow', fontsize=12, weight='bold')

    ax.set_title(title)
    ax.axis("off")
    plt.tight_layout()
    plt.show()

# Visualise cluster bounding boxes overlaid on the original image
draw_cluster_boundaries(
    transformed_img_np, agg_labels,
    title="Mean-Shift Cluster Boundaries on Transformed Image"
)
```

## Section 9: Single-Concept Ablation Study

For each causal concept (cluster), zero out its bounding box on the image and re-run inference. Comparing the modified prediction to the baseline reveals each concept's individual contribution to the model's decision.

```python
def zero_bbox(image_np, bbox):
    """
    Set all pixels inside a bounding box to zero (black) on a copy of the image.

    Args:
        image_np : [H, W, C]
        bbox     : Tuple (min_row, min_col, max_row, max_col)

    Returns:
        Modified copy with the specified region zeroed out
    """
    modified_image = image_np.copy()
    modified_image[bbox[0]:bbox[2], bbox[1]:bbox[3], :] = 0     # Set region to black
    return modified_image

# Print baseline prediction as reference point
print(f"Baseline Prediction: Class {prediction} with probability {predicted_prob:.4f}")
```

```python
unique_clusters = np.unique(agg_labels)  # All cluster ids including background (-1)
results = []                              # Store (cluster_id, prediction, probability) tuples

# Convert float [0,1] image to uint8 [0,255] for PIL compatibility
original_np = (transformed_img_np * 255).astype(np.uint8)

for cl in unique_clusters:
    if cl == -1:          # Skip background pixels
        continue

    # Compute bounding box for this cluster
    coords = np.argwhere(agg_labels == cl)
    min_row, min_col = coords.min(axis=0)
    max_row, max_col = coords.max(axis=0)
    bbox = (min_row, min_col, max_row + 1, max_col + 1)  # +1 for inclusive slicing

    # Create ablated image: zero out this concept's bounding box
    modified_np  = zero_bbox(original_np, bbox)
    modified_img = Image.fromarray(modified_np)

    # Pre-process and run inference on the ablated image
    input_tensor_modified = transform(modified_img).unsqueeze(0)
    input_tensor_modified.requires_grad = True

    output_modified      = model(input_tensor_modified)
    probabilities_modified = torch.softmax(output_modified, dim=1)
    prediction_modified  = torch.argmax(output_modified).item()
    predicted_prob_modified = probabilities_modified[0, prediction_modified].item()

    results.append((cl, prediction_modified, predicted_prob_modified))

    # Show the ablated image and its new prediction
    plt.figure(figsize=(6, 6))
    plt.imshow(modified_np)
    plt.title(
        f"Causal Concept {cl} Zeroed Out:\n"
        f"Predicted Class {prediction_modified} (Prob: {predicted_prob_modified:.4f})"
    )
    plt.axis("off")
    plt.show()

# Summarise ablation results vs. the baseline
print("\nComparison with Baseline:")
for cl, pred, prob in results:
    print(
        f"Causal Concept {cl}: Modified → Class {pred} (Prob: {prob:.4f}) "
        f"| Baseline → Class {prediction} (Prob: {predicted_prob:.4f})"
    )
```

## Section 10: Quantitative Metrics — PCS & Bayes Factor

Define and compute two metrics that quantify how much each causal concept influences the model's prediction:

- **PCS (Probability Change Score):** Relative drop in the baseline class probability.
- **BF (Bayes Factor):** Ratio of the model's confidence before vs. after removing the concept.

```python
def compute_pcs_and_bf(p_orig, p_mod, epsilon=1e-6):
    """
    Compute Probability Change Score (PCS) and Bayes Factor (BF).

    Args:
        p_orig  : Baseline class probability (before concept removal)
        p_mod   : Modified class probability (after concept removal)
        epsilon : Small constant to avoid division by zero / log(0)

    Returns:
        pcs (float) : Relative probability drop = (p_orig - p_mod) / p_orig
                      Positive → concept was helping the prediction
        bf  (float) : Odds ratio = [p_orig/(1-p_orig)] / [p_mod/(1-p_mod)]
                      >1 → model was more confident before removal
    """
    # Clip probabilities away from 0 and 1 to prevent numerical issues
    p_orig = np.clip(p_orig, epsilon, 1 - epsilon)
    p_mod  = np.clip(p_mod,  epsilon, 1 - epsilon)

    # Probability Change Score: normalised probability drop
    pcs = (p_orig - p_mod) / p_orig

    # Bayes Factor: ratio of log-odds (how much more probable original vs. modified)
    bf = (p_orig / (1 - p_orig)) / (p_mod / (1 - p_mod))

    return pcs, bf
```

## Section 11: Concept Summary Bar Chart

Visualise the probability change caused by removing each concept. Green bars = concept was *helping* the prediction; red bars = concept was *hurting* it.

```python
def plot_concept_summary(input_tensor, agg_labels, model, transform, prediction, predicted_prob):
    """
    Bar chart showing the change in baseline class probability when each
    causal concept's bounding box is zeroed out.

    Green bar (diff < 0): removing the concept lowered confidence: positive contribution
    Red   bar (diff > 0): removing the concept raised confidence: negative contribution
    """
    mean = [0.485, 0.456, 0.406]
    std  = [0.229, 0.224, 0.225]
    original_np    = unnormalize_and_to_np(input_tensor[0], mean, std)
    baseline_class = prediction
    baseline_prob  = predicted_prob

    # Helper: get bounding box for a given cluster label
    def get_bbox_for_cluster(cluster_label, label_map):
        coords = np.argwhere(label_map == cluster_label)
        min_row, min_col = coords.min(axis=0)
        max_row, max_col = coords.max(axis=0)
        return (min_row, min_col, max_row + 1, max_col + 1)

    clusters = [cl for cl in np.unique(agg_labels) if cl != -1]
    effects, concept_labels = [], []

    for cl in clusters:
        bbox       = get_bbox_for_cluster(cl, agg_labels)
        modified_np = zero_bbox(original_np, bbox)         # Ablate this concept
        modified_img = Image.fromarray(modified_np)
        input_mod  = transform(modified_img).unsqueeze(0)
        input_mod.requires_grad = True

        # Run inference on the ablated image
        output_mod = model(input_mod)
        probs_mod  = torch.softmax(output_mod, dim=1)
        mod_prob   = probs_mod[0, baseline_class].item()   # Probability of the original class

        effects.append(mod_prob - baseline_prob)           # Negative = probability dropped
        concept_labels.append(cl)

    # Colour: green for concepts that helped (removing them dropped probability)
    colors = ['green' if diff < 0 else 'red' for diff in effects]

    plt.figure(figsize=(10, 6))
    bars = plt.bar([str(c) for c in concept_labels], effects, color=colors)
    plt.xlabel("Causal Concept")
    plt.ylabel("Change in Baseline Class Probability\n(Modified − Baseline)")
    plt.title("Effect of Removing Each Causal Concept on Model Prediction")
    plt.axhline(0, color='black', linewidth=0.8)  # Reference line at zero

    # Annotate each bar with its numeric value
    for bar, diff in zip(bars, effects):
        plt.annotate(
            f"{diff:.3f}",
            xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
            xytext=(0, 3), textcoords="offset points",
            ha='center', va='bottom'
        )
    plt.show()

    # Text summary
    print("Summary of Causal Concept Effects (per concept):")
    for cl, diff in zip(concept_labels, effects):
        effect_type = "Positive Contribution" if diff < 0 else "Negative Effect"
        print(f"Concept {cl}: Change = {diff:.4f} → {effect_type}")

# Run the summary visualisation
plot_concept_summary(
    input_tensor=input_tensor, agg_labels=agg_labels,
    model=model, transform=transform,
    prediction=prediction, predicted_prob=predicted_prob
)
```

## Section 12: PCS Visualisation — Bounding Boxes Coloured by Importance

Draw bounding boxes on the original image, coloured by their PCS score. Darker red = higher PCS = more causally important concept.

```python
def plot_pcs_visualization(
    input_tensor, agg_labels, model, transform,
    prediction, predicted_prob, compute_pcs_and_bf
):
    """
    Overlay bounding boxes on the original image, colour-coded by PCS.
    Each box is annotated with its cluster id, PCS, and Bayes Factor.

    Reds colormap: darker = higher PCS = concept is more causally important.
    """
    mean = [0.485, 0.456, 0.406]
    std  = [0.229, 0.224, 0.225]
    original_np = unnormalize_and_to_np(input_tensor[0], mean, std)

    # Helper: bounding box for one cluster
    def get_bbox(cluster_label):
        coords = np.argwhere(agg_labels == cluster_label)
        min_row, min_col = coords.min(axis=0)
        max_row, max_col = coords.max(axis=0)
        return (min_row, min_col, max_row + 1, max_col + 1)

    detailed_results = []
    clusters = [cl for cl in np.unique(agg_labels) if cl != -1]

    for cl in clusters:
        bbox        = get_bbox(cl)
        modified_np = zero_bbox(original_np, bbox)
        modified_img = Image.fromarray(modified_np)
        input_mod   = transform(modified_img).unsqueeze(0)
        input_mod.requires_grad = True

        # Inference on ablated image
        out   = model(input_mod)
        probs = torch.softmax(out, dim=1)
        mod_prob  = probs[0, prediction].item()     # Prob of original class after ablation
        mod_pred  = torch.argmax(probs).item()

        pcs, bf = compute_pcs_and_bf(predicted_prob, mod_prob)

        detailed_results.append({
            'cluster': cl, 'modified_pred': mod_pred,
            'modified_prob': mod_prob,
            'pcs': pcs, 'bf': bf, 'bbox': bbox
        })

    # Sort by PCS descending so the most important concept is drawn first
    detailed_results.sort(key=lambda x: x['pcs'], reverse=True)

    fig, ax = plt.subplots(figsize=(8, 8))
    ax.imshow(original_np)
    ax.axis("off")

    # Map PCS values to colours using a Reds colormap
    cmap_r = cm.get_cmap("Reds")
    norm   = Normalize(vmin=0, vmax=max(r['pcs'] for r in detailed_results))

    for res in detailed_results:
        cl, pcs, bf, bbox = res['cluster'], res['pcs'], res['bf'], res['bbox']
        color = cmap_r(norm(pcs))  # Higher PCS → darker red

        rect = Rectangle(
            (bbox[1], bbox[0]),             # (x, y) = (col, row)
            bbox[3] - bbox[1],              # width
            bbox[2] - bbox[0],              # height
            linewidth=2, edgecolor=color, facecolor='none'
        )
        ax.add_patch(rect)
        ax.text(
            bbox[1], bbox[0] - 5,
            f"#{cl} | PCS: {pcs:.2f} | BF: {bf:.1f}",
            color='black', fontsize=9,
            bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round,pad=0.3')
        )
    plt.show()

# Run the final PCS visualisation
plot_pcs_visualization(
    input_tensor=input_tensor, agg_labels=agg_labels,
    model=model, transform=transform,
    prediction=prediction, predicted_prob=predicted_prob,
    compute_pcs_and_bf=compute_pcs_and_bf
)
```
